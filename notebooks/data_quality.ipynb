### Respuestas – Momento 3

**¿Qué diferencias existen entre int64, float64 y object?**
- `int64` guarda números enteros, sin decimales (por ejemplo `order_id` o `order_hour_of_day`). No admite valores ausentes.
- `float64` guarda números con decimales y sí admite `NaN`. Por eso `days_since_prior_order` y `add_to_cart_order`, aunque contienen enteros, aparecen como `float64`.
- `object` es el tipo genérico que las versiones anteriores de pandas usaban para el texto. En pandas 3 el texto aparece como `str` (es el caso de `product_name`, `aisle` y `department`).

**¿Qué información proporciona info() que no se obtiene directamente con shape?**
`shape` solo da el número de filas y de columnas. `info()` muestra además el nombre de cada columna, su tipo de dato, cuántos valores no nulos tiene (lo que revela los ausentes, por ejemplo 450 148 de 478 967 en `days_since_prior_order`) y la memoria que ocupa el DataFrame.

**¿En qué columnas se observan valores ausentes?**
| Archivo | Columna | Valores ausentes |
|---|---|---|
| orders | `days_since_prior_order` | 28 819 |
| products | `product_name` | 1 258 |
| order_products | `add_to_cart_order` | 836 |

Las columnas de `aisles` y `departments` no tienen ausentes.

**¿Qué posibles dificultades podrían generar estos valores en un análisis posterior?**
- **Promedios:** `mean()` ignora los `NaN`, así que el promedio de días entre pedidos se calcula solo con 450 148 valores y no con las 478 967 filas. Hay que decirlo al reportarlo.
- **Filtros:** una condición como `days_since_prior_order > 15` deja fuera las filas con `NaN`, porque una comparación con `NaN` siempre da falso. Un porcentaje calculado sobre el total de filas sale menor que uno calculado sobre los valores disponibles.
- **Productos sin nombre:** los 1 258 productos sin `product_name` no se pueden identificar en reportes (por ejemplo, en un "top" de productos más vendidos).
- **Cambio de tipo:** los `NaN` convierten columnas enteras en `float64`. Esto cambia cómo se muestran (`17.0`) y obliga a tratar los ausentes antes de convertirlas a `int64`.

### Filtros combinados

Se unen dos condiciones con `&` (y). Cada condición va entre paréntesis.

````python
comb1 = orders.loc[(orders["order_hour_of_day"] == 10) & (orders["order_number"] > 10)]

print("Solo hora 10 (ej3):", ej3.shape[0])
print("Solo order_number > 10 (ej5):", ej5.shape[0])
print("Ambas condiciones:", comb1.shape[0])
print(f"Porcentaje sobre el total de registros de pedidos: {comb1.shape[0] / total_orders * 100:.2f} %")
comb1.head()
````

**Explicación:** solo **21 537** registros (**4,50 %**) cumplen las dos condiciones a la vez. Es menos que cualquiera de los filtros por separado (40 578 de la hora 10 y 244 072 con `order_number > 10`): al exigir ambas condiciones, una fila debe pasar los dos filtros, por lo que el resultado nunca puede ser mayor que el filtro más pequeño.

````python
comb2 = products.loc[(products["department_id"] == 1) & (products["aisle_id"] > 50)]

print("Solo departamento 1 (ej2):", ej2.shape[0])
print("Ambas condiciones:", comb2.shape[0])
print(f"Porcentaje sobre el total de productos: {comb2.shape[0] / total_products * 100:.2f} %")
comb2.head()
````

**Explicación:** de los **4 008** productos del departamento 1 (congelados), **1 618** están además en un pasillo con `aisle_id` mayor que 50. Eso es el **3,26 %** del catálogo. De nuevo, la segunda condición reduce el resultado del filtro individual.

### Respuestas – Momento 4

**¿Qué diferencia existe entre seleccionar columnas y filtrar filas?**
Seleccionar columnas elige **qué información** se ve de cada registro sin cambiar cuántos hay: el ejercicio 1 mantuvo las 49 694 filas y redujo las columnas a 2. Filtrar filas elige **qué registros** se conservan según una condición, sin cambiar las columnas: el ejercicio 2 dejó 4 008 de 49 694 productos.

**¿Qué ventajas ofrece loc[] al consultar información específica?**
Permite indicar en una sola instrucción las filas y las columnas (`df.loc[filas, columnas]`), usar condiciones lógicas para las filas y nombres de columnas en lugar de posiciones numéricas, lo que hace el código más fácil de leer. Además devuelve un DataFrame que se puede guardar en una variable (`ej2`, `ej3`…) y volver a usar.

**¿Qué ocurre con la cantidad de registros cuando se aplican varias condiciones?**
Con `&` el resultado se reduce, porque cada fila debe cumplir todas las condiciones: 40 578 pedidos a las 10 h y 244 072 con `order_number > 10` se convierten en 21 537 cuando se exigen ambas. Con `|` (o) ocurriría lo contrario: el resultado sería igual o mayor que cada filtro por separado.

**¿Por qué es importante utilizar correctamente los operadores lógicos?**
- En pandas se usan `&` (y), `|` (o) y `~` (no), que comparan fila por fila. Las palabras `and`/`or` de Python esperan un solo valor verdadero o falso y producen un `ValueError` con una columna completa.
- Cada condición debe ir **entre paréntesis**, porque `&` y `|` se evalúan antes que `==` o `>`. Sin paréntesis el filtro falla o da un resultado distinto al esperado.
- Confundir `&` con `|` cambia por completo la respuesta (intersección frente a unión), y el error puede pasar inadvertido si no se comparan los conteos con los filtros individuales.
